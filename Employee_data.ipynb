```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["department"] = df["department"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

```python
df["salary"] = (
    df["salary"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(" INR", "", regex=False)
)

df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
```

```python
print(df.isnull().sum())
```

```python
df["salary"] = df["salary"].fillna(df["salary"].median())
```

```python
print(df.isnull().sum())
```

```python
df.loc[(df["age"] < 18) |(df["age"] > 65),"age"] = np.nan 
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["age"] = df["age"].astype(int)
```

```python
print(df.info())
```

```python
df["department"].value_counts(dropna=False)
```

```python
df["department"] = df["department"].str.strip().str.lower()

df["department"] = df["department"].replace({
    "it": "IT",
    "finance": "Finance",
    "h.r": "HR",
    "hr": "HR",
    "human resources": "HR",
    "sales": "Sales",
    "marketing": "Marketing",
    "operations": "Operations"
})
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.barplot(data=df, x="department", y="salary")
plt.title("Average Salary by Department")
plt.show()
```

```python
sns.countplot(data=df, x="department")
plt.title("Number of Employees by Department")
plt.show()
```

```python
role_salary = df.groupby("job_role")["salary"].median().sort_values(ascending=False)

sns.barplot(x=role_salary.values, y=role_salary.index)

plt.title("Average Salary by Job Role")
plt.xlabel("Average Salary")
plt.ylabel("Job Role")
plt.show()
```

```python
sns.histplot(df["age"], bins=16, kde=True)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

```python
sns.countplot(data=df, x="performance_score")

plt.title("Employee Performance Distribution")
plt.xlabel("Performance Score")
plt.ylabel("Employee Count")
plt.show()
```
